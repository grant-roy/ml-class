## Predicting probability of adoption given categorical input variables. 

The PDF accompanying this exercise outlines an example of logistic regression using 3 predictors.

Our predictors are:

* x1  : 0 if no college, 1 if some college or above.
* x2  : 0 if no change in residence in previous 5 years, 1 otherwise
* x3  : 0 if income is low, 1 if income is high

The *y* value, or outcome, is binary, 0 is they did not adopt, and 1 if they did adopt.  Your challenge will be to implement this example in python using the available information provided in the pdf. The following are the steps you will need to take:

* a) You must first create the dataset based on the specification given in the pdf. 
* b) Given the created dataset, you must set up the data in a way that allows you to take the dot product
     of the weights and your set of x predictors(don't forget to add a scalar term)
* c) you must use the logistic function to 'tame' the output of your dot product to a range between 0 and 1. 


Logistic function: 

$$ \frac {e^{w*x}}{1 + e^{w*x}} $$
