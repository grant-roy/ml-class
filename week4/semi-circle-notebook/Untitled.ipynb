## A closer look at random numbers

Sampling, and in general using a random number generator for simulations is an extremely powerful techinique that is very useful throughout probability and statistics.  Some of it's power in fact can often be suprising, beautiful, and elegant(contrary to what many analytic purists may think).

An example below will help solve this question: 

> How can we calculate the value of pi?

As it turns out we can use an rng to help us solve this. The way we do it, I hope, will make you smile. 


We will start will the idea of a unit circle(radius=1),and it's bounding rectangle(area=1). The ratio of Area of Circle/Area of Square  = pi*r^2/(2r)^2 = pi/4

Computing the ratio of the area of the circle to the area of the square, multiplied by 4 would give us a value for pi.

In essense we 'shoot points at the plane' and see how many end up inside the circle, vs the total number of points. This ratio is in fact pi.

### Numerical implementation of Monte Carlo Pi

Below is some code that will approximate pi via monte carlo simulation. The larger n value we give the more accurate
our estimate will be. 

```python
import random
from random import randint


def monte_carlo_pi(n):
    inside=0
    for point in range(n):
        x,y = random.random(), random.random()
        if(x*x + y*y < 1.0):
            inside+=1
    return 4.0*inside/n


print monte_carlo_pi(10000)
```
