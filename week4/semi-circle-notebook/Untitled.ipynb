We will create two semi circular regions, and then sample 1000 random points from each. We will run first the perceptron and then a linear regression. 


### Perceptron 

```python
import semi_circles as sc
import perceptron as pt
import scatter_plot as sp


%matplotlib inline

samples  = sc.semi_circular_samples(offset=7.5,sep=5)
w = pt.perceptron(samples)
sp.scatter(samples, w)
```

### Regression

```python
import regression as rg

data  = samples[['x','y']]
m,c  = rg.regression(data)

rg.plot_regression(data, m, c)
```
