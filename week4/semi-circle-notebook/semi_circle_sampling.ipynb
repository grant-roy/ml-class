## A closer look at random numbers

Sampling, and in general using a random number generator for simulations is an extremely powerful techinique that is very useful throughout probability and statistics.  Some of it's power in fact can often be suprising, beautiful, and elegant(contrary to what many analytic purists may think).

An example below will help solve this question: 

> How can we calculate the value of pi?

As it turns out we can use an rng to help us solve this. The way we do it, I hope, will make you smile. 

We will start will the idea of a unit circle($r=1$), and it's bounding rectangle($area=4$). The ratio of 

$$ \begin{align} 
\frac{Area of Circle}{Area of Rectangle}& = \frac{ \pi r^2}{(2r)^2} \\  
& = \frac{ \pi }{4}  \\
& = \frac {4 \pi}{4} & \text{multiply by 4} \\
& = \pi 
\end{align}
$$
 
Thus computing the ratio of the area of the circle to the area of the square, and then multiplying by 4 would give us an approximate value for pi.

In essense we 'shoot points at the plane' and see how many end up inside the circle, vs the total number of points. This ratio is in fact approximately pi. 

Below is a visualization of what this process looks like, note that the circle may appear as an elipse, but this is an artifact of plot perspective/scale...it is indeed a unit circle #realife


```python
import calculate_pi as cp
%matplotlib inline

cp.plot_unit_circle(1000) #  we throw 1000 'darts'
```

## Probability with Monte Carlo

Let's take a look at an example where we can use Monte Carlo simulation to estimate an actual probability. In this case we have an empirical answer to compare it to. This is an example problem from the book *Digital Dice* by Paul J. Nahin, which I can strongly recommend. Where the book uses Matlab code, the following is of course an implementation in python. 

## The problem

> Clumsy Dishwasher

There are five dishwashers in a kitchen, 5 dishes are broken on a given day. Dishes are *handled equally by them all under equal circumstances*, and one of the dishwashers broke 4. The dishwasher who broke 4 gives the excuse that 'it could have happened to anyone, I'm not really clumsy!'. 

What is the probability that the dishwasher who broke 4 dishes is not clumsy? 

Analytically we can give the solution as: 

$$ \frac{4 {5 \choose 4} + {5 \choose 5} }{5^5} = \frac{20+1}{3,125} = .00672 $$

Computationally we can solve this problem with a Monte Carlo simulation and verify our analytic answer

```python
import dishwasher as dw


print dw.probability_of_breaking(10000000)
```
