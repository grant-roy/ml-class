## Bias Variance Analysis

Here we will present the code for the two examples in the book,

We consider a function 

$$ f(x) = sin(\pi x) $$

$x$ in the formula above will be the closed interval from [-1,1]. 

### Horizontal line through the midpoint

$H_0: $ Set of all lines of the form $h(x) = b$

Sample randomly two points that lie on the line representing the graph of the function on it's domain. These will be $(x_1,y_1)$ and $(x_2,y_2)$

Now we will form our hypotheses by doing

$$ h(x) = b = \frac{y_1 + y_2}{2} $$

We will do this sampling of points many times to come up with a series of different hypotheses. Take note of how we are varying the *data* we feed into our algorithm by sampling different random points.

The code below shows a graph of what this looks like



```python
import bias_variance as bv
%matplotlib inline

fits = bv.fit_fx()
bv.plot(target=fits['target'], y_lines=fits['y_lines'])
```

## Calculating the actual bias and variance

So we have a function, plotted in blue above that is $f(x) = sin(\pi x)$ over the $x$ domain of $[-1,1]$ in steps of $.01$, which is why you see 200 along the $x$ axis, that is there are $200$ steps of $.01$ from $-1:1$ along the $x$ axis.

Each one of the red lines above was created from a point, and that point was calculated as a result of the two random points we chose along the blue line. So to create a red line, we chose two random points that fall on the blue line, and we take both of their $y \ values$ to calulcate:

$$x = size \ of \ dataset = number \ of \ ramdom \ samples = 2$$

$$h(x) = b = \frac{y_1+y_2}{ 2}$$

giving us a $g(x)= b$. This is effectively the $midpoint$ between the $y_1,y_2$ we have sampled.  We will repeat this process many times, giving us an array with many $g(x)$ values, however many times we chose to randomly sample two points along the blue line. We will denote this array containing many g(x) estimates the set of $g(x)$:
$$\{g(x)\} =  \ set \ of \ all \ g(x)$$ 

And we will also denote the length of $\{g(x)\}$ as $N_{g(x)}$

> $\bar{g}$ calculation

We first need to calculate $\bar{g}$, which is the average value of $\{g(x)\}$ : 

$$ \bar{g} = \frac{1}{N_{g(x)}} \sum{g(x)} $$

In words: the above is a simple average: add up all the values in $\{g(x)\}$ and divide by the number of items in $\{g(x)\}$.

> Bias Calculation

To calculate the bias, what we are going to do is essentially is measure the distance between the average hypothesis $\bar{g}$ and the function $f(x) = sin(\pi x)$ for every $x$ value we have in the interval [-1,1].  We then take the average of that value by dividing it by how many values there are in the interval [-1, 1]. 

So basically this is a least squares fit of the average hypotheses $\bar{g}$ to our target function $f(x)$, where we then average that value by the number of points in $f(x)$

We will denote the number of points in our target function $f(x)$ as $N_{f(x)}$

We then have 

$$ bias  = \frac{1}{N_{f(x)}} \sum{(\bar{g} - f(x))}{^2}$$

In words: subtract $\bar{g}$ from every value in $f(x)$, square the resulting value, and then add all of those value up. Take the average of this final value by dividing it by the total number of points in $f(x)$

> Variance calculation

To calculate the variance we will again need the quantity $\bar{g}$ that we calculated above. This time however instead of focusing our attention on the difference between $\bar(g)$ and our target function $f(x)$, we will look at the difference between $\bar{g}$ and the other hypotheses in $\{g(x)\}$. This is effectively a measure of how much the average hypotheses differs from others, in other words, is it very difference from all the others? This would indicate that the values in $\{g(x)\}$ do $vary$ quite a bit. Hence this is a measure of the variance of the hypotheses that our model produces. 

To calculate variance we have

$$ variance = \frac{1}{N_{g(x)}} \sum{(\bar{g} - g(x))}{^2}$$

```python
bias = bv.calculate_bias(target=fits['target'], g_x=fits['g_x'])
variance = bv.calculate_variance(g_x=fits['g_x'])

print bias
print variance
```

### Line that passes through the two data points

$H_1: $ Set of all lines of the form $h(x) = ax + b$  

Our hypothesis is a bit more complicated this time as it is not a *point*, but rather it is the equation of the line through our two sample points $(x_1,y_1), (x_2,y_2)$

Our calculation will look like this

$$ \forall x_i \ in \ [-1,1]; \ \  h(x) = \frac{y_2 - y_1}{x_2-x_1} \times (x_i-x_1) \times b $$ 

where $b$ is the $y \ intercept$ given by the formula for the equation of a line. 

This will give us an array **for each sample we take**. In this case ${g(x)}$ will be an array of composed of arrays representing each hyposthesis line. How do we do this calculation with an array?

```python

fits = bv.fit_fx(horizontal=False)
bias = bv.calculate_bias(target=fits['target'], g_x=fits['g_x'], h1=True)
variance = bv.calculate_variance(g_x=fits['g_x'])


bv.plot(target=fits['target'], y_lines=fits['y_lines'])
print bias
print variance
```
