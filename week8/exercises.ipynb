## Exercises working with the bias variance calculation

We will look at Example 2.8 from page 65 of the book


*  **a)** Plot the graph of the sin function on the x interval [-1, 1]. This function is given by $ f(x) = sin(\pi x)$ 

*  **b)** Sample two points that lie on the sin curve formed by $f(x)$.  Calculate 

$$ \mathbf{H}_0:   h(x) = \frac{y_1 + y_2}{2} $$

 > plot the resulting line formed by $y = h(x)$

*  **c)** Now we will repeat 2. above many times, sampling 500 points that lie on the sin curve, calculating an h(x) for each one. Store all of the resulting hypothesis in an array, this array will form $g(x)$. 

*  **d)** Calculate the bias 

$$ bias  = \frac{1}{N_{f(x)}} \sum{(\bar{g} - f(x))}{^2}$$

where

$$ \bar{g} = \frac{1}{N_{g(x)}} \sum{g(x)} $$

*  **e)** Calculate the variance 

$$ variance = \frac{1}{N_{g(x)}} \sum{(\bar{g} - g(x))}{^2}$$
